#### Feature Engineering

```python
# Creating new variables for the dataset

# Fighters dataset:

# Convert columns to numeric types
fighters['reach'] = pd.to_numeric(fighters['reach'], errors='coerce')
fighters['height'] = pd.to_numeric(fighters['height'], errors='coerce')
fighters['age'] = pd.to_numeric(fighters['age'], errors='coerce')

# Calculate differences
fighters['reach_diff'] = fighters['reach'] - fighters['reach'].shift(1)
fighters['height_diff'] = fighters['height'] - fighters['height'].shift(1)
fighters['age_diff'] = fighters['age'] - fighters['age'].shift(1)


# Matches dataset:
matches['age_diff'] = matches['age_A'] - matches['age_B']
matches['reach_diff'] = matches['reach_A'] - matches['reach_B']
matches['height_diff'] = matches['height_A'] - matches['height_B']
matches['weight_diff'] = matches['weight_A'] - matches['weight_B']
```

```python
# for now fill the null values with 0
# copy matches dataset to new dataset
matches_null_is_zero = matches.copy()
matches_null_is_zero.fillna(0, inplace=True)
```
